RandomForest / DecisionTree / NaiveBayes を 10-fold CV で比較する。

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, numpy as np
from src.classification.classifier import evaluate, MODELS
from src.features.feature_merger import select_by_pearson
```

```python
df = pd.read_csv('../data/processed/features.csv')
y = df['label'].values
X_df = df.drop(columns=['label']).fillna(0)

# Pearson 特徴選択
selected = select_by_pearson(X_df, y, threshold=0.1)
print(f'selected {len(selected)} features')
X = X_df[selected].values
```

```python
results = {}
for name in MODELS:
    results[name] = evaluate(X, y, name)
    print(name, results[name])
pd.DataFrame(results).T
```
