# SHAP 特徴重要度分析

最終モデルの特徴重要度を可視化し、自然言語で判定理由を出力する。

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, numpy as np
from sklearn.ensemble import RandomForestClassifier
from src.explainability.shap_analyzer import compute_shap_values, plot_summary, generate_explanation
from src.features.feature_merger import select_by_pearson
```

```python
df = pd.read_csv('../data/processed/features.csv')
y = df['label'].values
X_df = df.drop(columns=['label']).fillna(0)
selected = select_by_pearson(X_df, y, threshold=0.1)
X = X_df[selected].values

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

shap_values = compute_shap_values(model, X)
plot_summary(shap_values[1], X_df[selected])
```

```python
# サンプル1件の説明文を生成
print(generate_explanation(shap_values[1][0], selected))
```
